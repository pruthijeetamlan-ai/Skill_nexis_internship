```python
import pandas as pd 
import numpy as np
```

```python
tit = pd.read_csv(r"C:\Users\pruth\Downloads\archive\Titanic-Dataset.csv")
```

```python
tit.info()
```

```python
tit.describe()
```

```python
tit.isnull().sum()
```

```python
# Mean imputation
tit['Age'] = tit['Age'].fillna(tit['Age'].mean())
print("Mean Age:", tit['Age'].mean())
```

```python
# Median imputation

tit['Age'] = tit['Age'].fillna(tit['Age'].median())
print("Median Age:", tit['Age'].median())
```

```python
# Mode imputation
tit['Embarked'] = tit['Embarked'].fillna(tit['Embarked'].mode()[0])
print(tit.mode().iloc[0])
```

```python
print(tit.isnull().sum())
```

```python
# Encode

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
tit['Sex'] = le.fit_transform(tit['Sex'])
```

```python

#One-Hot Encoding
tit = pd.get_dummies(tit, columns=['Embarked'], dtype=int)
```

```python
print(tit.head())
print(tit.info())
```

```python
print(tit.isnull().sum())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.histplot(data=tit, x='Age', bins=20, kde=True)

plt.title('Age Distribution of Titanic Passengers')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
tit.to_csv('titanic_cleaned.csv', index=False)
print("Cleaned dataset saved successfully!")
```
